```python
import numpy as np
import matplotlib.pyplot as plt

actual = np.load("actual.npy")
lstm = np.load("lstm_preds.npy")
transformer = np.load("transformer_preds.npy")
```

```python
actual = actual.reshape(-1)
lstm = lstm.reshape(-1)
transformer = transformer.reshape(-1)
```

```python
plt.figure(figsize=(15,6))

plt.plot(actual[:500], label="Actual", linewidth=2)

plt.plot(lstm[:500], label="LSTM Prediction")

plt.plot(transformer[:500], label="Transformer Prediction")

plt.xlabel("Time Step")
plt.ylabel("Temperature")
plt.title("Actual vs Predicted Temperature")

plt.legend()
plt.grid(True)

plt.show()
```
